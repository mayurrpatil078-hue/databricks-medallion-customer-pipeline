```python

from pyspark.sql import functions as F

data = [
    (101, "Amit",   "Pune",   "amit@example.com",   "2026-10-01"),
    (102, "Priya",  "Mumbai", "priya@example.com",  "2026-10-02"),
    (103, "Rahul",  "Nashik", None,                 "2026-10-03"),
    (104, "Sneha",  "Pune",   "sneha@example.com",  "2026-10-04"),
    (105, "Vijay",  "Nagpur", "vijay@example.com",  "2026-10-05")
]

columns = [
    "customer_id",
    "customer_name",
    "city",
    "email",
    "source_updated_date"
]

df = spark.createDataFrame(data, columns)

df = df.withColumn(
    "source_updated_date",
    F.to_date("source_updated_date")
)

display(df)
```

```python
catalog_name = "myadlasdevcalaog"

bronze_table = f"{catalog_name}.bronze.customers"

df = df.withColumn(
    "ingestion_timestamp",
    F.current_timestamp()
)

df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(bronze_table)

print(f"Bronze table created: {bronze_table}")
```
